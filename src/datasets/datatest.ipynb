```python
from scipy.io import loadmat
import numpy as np
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
def load_mat_file(filepath):
    try:
        data = loadmat(filepath, struct_as_record=False, squeeze_me=True)
        cleaned = {k: v for k, v in data.items() if not k.startswith("__")}
        return cleaned

    except NotImplementedError:
        import h5py

        with h5py.File(filepath, "r") as f:
            return {k: np.array(v) for k, v in f.items()}
```

```python
# # Burgers1D dataset information:
grid_size = 8192
mat_data = load_mat_file("./Burgers1D/burgers_data_R10.mat")
for var_name, var_value in mat_data.items():
    print(f"{var_name}: shape={var_value.shape}, dtype={var_value.dtype}")

# # data information
# # mat_data.keys() -> dict_keys(['a', 'a_smooth', 'a_smooth_x', 'a_x', 'u'])
# a: shape=(2048, 8192), dtype=float64
# a_smooth: shape=(2048, 8192), dtype=float64
# a_smooth_x: shape=(2048, 8191), dtype=float64
# a_x: shape=(2048, 8191), dtype=float64
# u: shape=(2048, 8192), dtype=float64
```

```python
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Times New Roman"]
plt.rcParams["axes.unicode_minus"] = False

for index in range(2):
    # coeff = mat_data["a"][index]  # initial / forcing coefficient along x
    coeff = mat_data["a_smooth"][index]
    sol = mat_data["u"][index]

    n = coeff.shape[-1]
    x = np.linspace(0.0, 1.0, n)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 6), sharex=True)

    ax1.plot(x, coeff, color="C0", linewidth=0.8)
    ax1.set_xlabel("$x$", fontsize=11)
    ax1.set_ylabel("$a(x)$", fontsize=11)
    ax1.set_title("Coefficient (1D Burgers)", fontsize=12)
    ax1.grid(True, alpha=0.3)

    ax2.plot(x, sol, color="C3", linewidth=0.8)
    ax2.set_xlabel("$x$", fontsize=11)
    ax2.set_ylabel("$u(x)$", fontsize=11)
    ax2.set_title("Solution (1D Burgers)", fontsize=12)
    ax2.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
```

```python
# Darcy2D dataset information:
grid_size = 241
mat_data = load_mat_file("./Darcy2D/piececonst_r241_N1024_smooth1.mat")
for var_name, var_value in mat_data.items():
    print(f"{var_name}: shape={var_value.shape}, dtype={var_value.dtype}")


# data information:
# mat_data.keys() -> dict_keys(['Kcoeff', 'Kcoeff_x', 'Kcoeff_y', 'coeff', 'sol'])
# Kcoeff: shape=(1024, 241, 241), dtype=float64
# Kcoeff_x: shape=(1024, 241, 241), dtype=float64
# Kcoeff_y: shape=(1024, 241, 241), dtype=float64
# coeff: shape=(1024, 241, 241), dtype=float32
# sol: shape=(1024, 241, 241), dtype=float32
```

```python
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Times New Roman"]
plt.rcParams["axes.unicode_minus"] = False

for index in range(2):
    coeff = mat_data["Kcoeff"][index]  # Kcoeff <- smoothed coeff
    # coeff = mat_data['coeff'][index] # shape coeff
    sol = mat_data["sol"][index]

    fig = plt.figure(figsize=(10, 6))

    ax1 = fig.add_subplot(121, projection="3d")
    X, Y = np.meshgrid(np.arange(grid_size), np.arange(grid_size))
    surf1 = ax1.plot_surface(
        X, Y, coeff, cmap="RdBu_r", linewidth=0, antialiased=True, alpha=0.95
    )
    ax1.set_xlabel("X", fontsize=11)
    ax1.set_ylabel("Y", fontsize=11)
    ax1.set_zlabel("coeff", fontsize=11)
    ax1.set_title("Coefficient Function", fontsize=12)
    ax1.set_zlim(0, 12)
    fig.colorbar(surf1, ax=ax1, shrink=0.5, aspect=15)

    ax2 = fig.add_subplot(122, projection="3d")
    surf2 = ax2.plot_surface(
        X, Y, sol, cmap="RdBu_r", linewidth=0, antialiased=True, alpha=0.95
    )
    ax2.set_xlabel("X", fontsize=11)
    ax2.set_ylabel("Y", fontsize=11)
    ax2.set_zlabel("u", fontsize=11)
    ax2.set_title("Solution Function", fontsize=12)
    ax2.set_zlim(0, 0.02)
    fig.colorbar(surf2, ax=ax2, shrink=0.5, aspect=15)

    plt.tight_layout()
    plt.show()
```
