```python
from scipy.io import loadmat
import numpy as np
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
def load_mat_file(filepath):
    try:
        data = loadmat(filepath, struct_as_record=False, squeeze_me=True)
        cleaned = {k: v for k, v in data.items() 
                  if not k.startswith('__')}
        return cleaned
        
    except NotImplementedError:
        import h5py
        with h5py.File(filepath, 'r') as f:
            return {k: np.array(v) for k, v in f.items()}
```

```python
grid_size = 241
mat_data = load_mat_file('Darcy2D\piececonst_r241_N1024_smooth1.mat')
for var_name, var_value in mat_data.items():
    print(f"{var_name}: shape={var_value.shape}, dtype={var_value.dtype}")
```

```python
# coeff是系数函数a(x,y) sol是解函数u(x,y) Kcoeff, Kcoeff_x, Kcoeff_y是什么
```

```python

plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman']
plt.rcParams['axes.unicode_minus'] = False

for index in range(5):
    coeff = mat_data['Kcoeff'][index]
    # coeff = mat_data['coeff'][index]
    sol = mat_data['sol'][index]

    fig = plt.figure(figsize=(10, 6))

    ax1 = fig.add_subplot(121, projection='3d')
    X, Y = np.meshgrid(np.arange(grid_size), np.arange(grid_size))
    surf1 = ax1.plot_surface(X, Y, coeff, cmap='RdBu_r', 
                            linewidth=0, antialiased=True, alpha=0.95)
    ax1.set_xlabel('X', fontsize=11)
    ax1.set_ylabel('Y', fontsize=11)
    ax1.set_zlabel('coeff', fontsize=11)
    ax1.set_title('Coefficient Function', fontsize=12)
    ax1.set_zlim(0, 12)
    fig.colorbar(surf1, ax=ax1, shrink=0.5, aspect=15)

    ax2 = fig.add_subplot(122, projection='3d')
    surf2 = ax2.plot_surface(X, Y, sol, cmap='RdBu_r', 
                            linewidth=0, antialiased=True, alpha=0.95)
    ax2.set_xlabel('X', fontsize=11)
    ax2.set_ylabel('Y', fontsize=11)
    ax2.set_zlabel('u', fontsize=11)
    ax2.set_title('Solution Function', fontsize=12)
    ax2.set_zlim(0, 0.02)
    fig.colorbar(surf2, ax=ax2, shrink=0.5, aspect=15)
    
    plt.tight_layout()
    plt.show()
```
